```python
# ==========================================
# TAHAP 1: IMPORT LIBRARY
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Pengaturan tampilan grafik
sns.set_theme(style="whitegrid")

print("Semua library berhasil diimpor!")
```

```python
# ==========================================
# TAHAP 2: MEMBACA DATASET EXCEL
# ==========================================

file_path = "Dataset_Pasien_Data_Cleansing.xlsx"

# Memeriksa nama sheet dalam Excel
excel = pd.ExcelFile(file_path)

print("Daftar sheet dalam dataset:")
print(excel.sheet_names)
```

```python
# ==========================================
# TAHAP 3: MEMBACA DATA PASIEN
# ==========================================

# Membaca sheet Dataset Mentah
df = pd.read_excel(
    file_path,
    sheet_name="Dataset Mentah"
)

# Menampilkan 5 baris pertama
print("Lima data pertama:")
display(df.head())

# Menampilkan jumlah baris dan kolom
print("\nUkuran dataset:")
print(f"Jumlah baris  : {df.shape[0]}")
print(f"Jumlah kolom  : {df.shape[1]}")

# Menampilkan nama kolom
print("\nNama kolom:")
print(df.columns.tolist())
```

```python
# ==========================================
# TAHAP 4: PEMERIKSAAN KUALITAS DATA
# ==========================================

# Informasi tipe data dan jumlah data terisi
print("Informasi dataset:")
df.info()

# Memeriksa nilai kosong
print("\nJumlah nilai kosong:")
display(df.isnull().sum())

# Memeriksa jumlah data duplikat
print("\nJumlah baris duplikat:")
print(df.duplicated().sum())

# Statistik deskriptif untuk kolom numerik
print("\nStatistik deskriptif:")
display(df.describe(include="all"))
```

```python
# ==========================================
# TAHAP 4: PEMERIKSAAN KUALITAS DATA
# ==========================================

# Informasi tipe data dan jumlah data terisi
print("Informasi dataset:")
df.info()

# Memeriksa nilai kosong
print("\nJumlah nilai kosong:")
display(df.isnull().sum())

# Memeriksa jumlah data duplikat
print("\nJumlah baris duplikat:")
print(df.duplicated().sum())

# Statistik deskriptif untuk kolom numerik
print("\nStatistik deskriptif:")
display(df.describe(include="all"))
```

```python
# ==========================================
# TAHAP 5: HISTOGRAM DISTRIBUSI UMUR
# ==========================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Umur",
    bins=8,
    kde=True
)

plt.title("Distribusi Umur Pasien", fontsize=16)
plt.xlabel("Umur Pasien (Tahun)")
plt.ylabel("Frekuensi Pasien")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 6: BAR CHART DIAGNOSIS
# ==========================================

plt.figure(figsize=(10, 6))

ax = sns.countplot(
    data=df,
    x="Diagnosis",
    order=df["Diagnosis"].value_counts().index
)

ax.bar_label(ax.containers[0], padding=3)

plt.title("Jumlah Pasien Berdasarkan Diagnosis", fontsize=16)
plt.xlabel("Kategori Diagnosis")
plt.ylabel("Jumlah Pasien")
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 7: PIE CHART STATUS RAWAT
# ==========================================

status_counts = df["Status Rawat"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    status_counts.values,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Persentase Pasien Berdasarkan Status Rawat", fontsize=15)
plt.legend(title="Status Rawat")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 8: BOXPLOT UMUR PASIEN
# ==========================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Status Rawat",
    y="Umur"
)

plt.title("Distribusi Umur Berdasarkan Status Rawat", fontsize=15)
plt.xlabel("Status Rawat")
plt.ylabel("Umur Pasien (Tahun)")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 9: STACKED BAR CHART
# ==========================================

tabel_diagnosis = pd.crosstab(
    df["Diagnosis"],
    df["Status Rawat"]
)

ax = tabel_diagnosis.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

plt.title("Diagnosis Pasien Berdasarkan Status Rawat", fontsize=15)
plt.xlabel("Kategori Diagnosis")
plt.ylabel("Jumlah Pasien")
plt.xticks(rotation=30, ha="right")
plt.legend(title="Status Rawat")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 10: HEATMAP
# ==========================================

plt.figure(figsize=(9, 6))

sns.heatmap(
    tabel_diagnosis,
    annot=True,
    fmt="d",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Heatmap Diagnosis dan Status Rawat", fontsize=15)
plt.xlabel("Status Rawat")
plt.ylabel("Diagnosis")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TAHAP 11: MENYIMPAN GRAFIK DAN DATA
# ==========================================

import os

folder_hasil = "hasil_visualisasi"
os.makedirs(folder_hasil, exist_ok=True)

# Simpan dataset yang sedang digunakan
df.to_csv(
    f"{folder_hasil}/Dataset_Pasien.csv",
    index=False
)

# Simpan tabel diagnosis dan status rawat
tabel_diagnosis.to_csv(
    f"{folder_hasil}/Tabel_Diagnosis_Status.csv"
)

print("Dataset dan tabel berhasil disimpan.")
print("Folder hasil:", folder_hasil)
```
